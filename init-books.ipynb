### Importação das Bibliotecas necessárias

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup as soup
import plotly.express as px
```

#### Conexão com site https://books.toscrape.com/

```python
url = 'https://books.toscrape.com/'
request = requests.get(url)
request
```

#### Scraping 

```python
site = soup(request.text, 'html')
site
```

## Tratamento de dados

```python
info_name = site.find_all('div', class_ = 'image_container')
display(info_name)
```

```python
book_name = []
for item in info_name:
    nome = item.a.img['alt']
    book_name.append(nome)
display(book_name)
```

```python
price_info = (site.find_all('p', class_ = 'price_color'))

price = []
for item in price_info:
    price.append(str(item.text))
for item in price:
    prices = list(map(lambda x: x.strip('Â£'), price))
print(prices)
```

## Criação de um DataFrame

```python
df = pd.DataFrame(book_name, columns=['book name'])
display(df)
```

```python
df['price'] = prices
display(df)
```

### Exportar o DataFrame pra CSV

```python
df.to_csv('books_info.cvs', index=False)
```

### Resumo dos dados com Gráfico

```python
fig = px.bar(df, x='book name', y='price', color='price', title='Preços de Livros')
fig.show()
```
